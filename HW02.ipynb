### Salt and Pepper Noise

```python
sp = np.random.randint(0,100,np.shape(f))
f_sp = f.copy()
f_sp[sp<3] = 0
f_sp[sp>96] = 255

plt.figure(dpi=128)
plt.imshow(f_sp,cmap='gray',aspect='equal')
plt.axis('off')
plt.show()
```

```python
plt.hist(f.ravel(),256)
plt.title('histogram of corrupted image(f+n)')
plt.show()

plt.hist(f_sp.ravel(),256)
plt.title('histogram of corrupted image(f+n)')
plt.show()
```

### Mean vs Median Filtering

```python
SS = 7 # kernel size (square kernel)
ss = int((SS-1)/2)
r_median = np.zeros(np.shape(f_sp))
r_mean = np.zeros(np.shape(f_sp))
```

```python
for x in range(ss,np.size(f_sp,0)-ss):
  for y in range(ss,np.size(f_sp,1)-ss):
    Sxy = f_sp[x-ss:x-ss+SS,y-ss:y-ss+SS]
    r_mean[x,y] = np.mean(Sxy)
    r_median[x,y] = np.median(Sxy)
```

```python
plt.figure(figsize=(16,9))
plt.subplot(131)
plt.imshow(f_sp,cmap='gray',aspect='equal')
plt.axis('off')
plt.title('corrupted image')
plt.subplot(132)
plt.imshow(r_mean,cmap='gray',aspect='equal')
plt.axis('off')
plt.title('mean filtered')
plt.subplot(133)
plt.imshow(r_median,cmap='gray',aspect='equal')
plt.axis('off')
plt.title('median filtered')
plt.show()
```

## Adaptive Local Noise Reduction

```python
# Adaptive local noise reduction
f = f_n

std_eta = 16
SS = 5

ss = int((SS-1)/2)
r_amean = np.zeros(np.shape(f))
for x in range(ss,np.size(f,0)-ss):
  for y in range(ss,np.size(f,1)-ss):
    Sxy = f[x-ss:x-ss+SS,y-ss:y-ss+SS]
    r_mean[x,y] = np.mean(Sxy)
    std_Sxy = np.std(Sxy)
    r_amean[x,y]=f[x,y]-((std_eta*std_eta)/(std_Sxy*std_Sxy))*(f[x,y]-np.mean(Sxy))
```

```python
plt.figure(figsize=(16,9))
plt.subplot(131)
plt.imshow(f_n,cmap='gray',aspect='equal',vmin=0,vmax=255)
plt.axis('off')
plt.title('corrupted image')
plt.subplot(132)
plt.imshow(r_mean,cmap='gray',aspect='equal',vmin=0,vmax=255)
plt.axis('off')
plt.title('mean filtered')
plt.subplot(133)
plt.imshow(r_amean,cmap='gray',aspect='equal',vmin=0,vmax=255)
plt.axis('off')
plt.title('adaptive local noise reduction')
plt.show()
```

## Adaptive Median Filtering

```python
# Adaptive Median Filtering
#f = f_sp
f = plt.imread(imgpath+'/circuitboard-saltandpep.tif')
SS_vals = np.array([3, 5, 7])
r_amedian = np.zeros(np.shape(f))
r_median = np.zeros(np.shape(f))
r_case = np.zeros(np.shape(f))

ss_max = int((np.max(SS_vals)-1)/2)
for x in range(ss_max,np.size(f,0)-ss_max):
  for y in range(ss_max,np.size(f,1)-ss_max):
    flag_go = 1
    cnt = 0
    z_xy = f[x,y]
    for SS in SS_vals:
      ss = int((SS-1)/2)
      Sxy = f[x-ss:x-ss+SS,y-ss:y-ss+SS]
      z_med = np.median(Sxy)
      z_min = np.min(Sxy)
      z_max = np.max(Sxy)

      if cnt==0:
        r_median[x,y] = z_med

      if z_min<z_med and z_med<z_max:
        if z_min<z_xy and z_xy<z_max:
          r_amedian[x,y] = z_xy
          flag_go = 0
        else:
          r_amedian[x,y] = z_med
          flag_go = 0
      else:
        if SS == np.max(SS_vals):
          r_amedian[x,y] = z_med
          flag_go = 0

      if flag_go==0:
        r_case[x,y] = cnt
        break
      cnt = cnt+1
```

```python
plt.figure(figsize=(16,9))
plt.subplot(131)
plt.imshow(f,cmap='gray',aspect='equal',vmin=0,vmax=255)
plt.axis('off')
plt.title('corrupted image')
plt.subplot(132)
plt.imshow(r_median,cmap='gray',aspect='equal',vmin=0,vmax=255)
plt.axis('off')
plt.title('median filtered')
plt.subplot(133)
plt.imshow(r_amedian,cmap='gray',aspect='equal',vmin=0,vmax=255)
plt.axis('off')
plt.title('adaptive median filtered')
plt.show()
```

```python
plt.imshow(r_case,cmap='jet',interpolation='nearest')
plt.axis('off')
plt.colorbar()
plt.show()
```

## Exercise
- 두 개의 noisy image를 각각 f1, f2로 불러오고 적절한 restoration(denoising)을 수행해보자.

```python
f1 = plt.imread(imgpath+'/galaxy-pair-noisy.tif')
f2 = plt.imread(imgpath+'/dentalXray-pepper-noise.tif')

plt.figure(dpi=128)
plt.subplot(121)
plt.imshow(f1,cmap='gray')
plt.axis('off')
plt.subplot(122)
plt.imshow(f2,cmap='gray')
plt.axis('off')
plt.show()
```

- 복원코드를 직접 작성해보기.

```python
# 1. Arithmetic Mean Filter
def arithmetic_mean_filter(img, kernel_size=3):
    ss = kernel_size // 2
    padded = np.pad(img, ss, mode='edge')
    result = np.zeros_like(img, dtype=np.float64)

    for x in range(img.shape[0]):
        for y in range(img.shape[1]):
            Sxy = padded[x : x + kernel_size, y : y + kernel_size]
            result[x, y] = np.mean(Sxy)
    return result

# 2. Median Filter
def median_filter(img, kernel_size=3):
    ss = kernel_size // 2
    padded = np.pad(img, ss, mode='edge')
    result = np.zeros_like(img, dtype=np.float64)

    for x in range(img.shape[0]):
        for y in range(img.shape[1]):
            Sxy = padded[x : x + kernel_size, y : y + kernel_size]
            result[x, y] = np.median(Sxy)
    return result

def adaptive_local_noise_reduction(img, kernel_size=5, noise_var=100):
    ss = kernel_size // 2
    padded = np.pad(img, ss, mode='edge')
    result = np.zeros_like(img, dtype=np.float64)

    for x in range(img.shape[0]):
        for y in range(img.shape[1]):
            Sxy = padded[x : x + kernel_size, y : y + kernel_size]
            local_mean = np.mean(Sxy)
            local_var = np.var(Sxy)

            g_xy = img[x, y]

            if local_var == 0:
                result[x, y] = g_xy
            else:
                ratio = min(1.0, noise_var / local_var)
                result[x, y] = g_xy - ratio * (g_xy - local_mean)
    return np.clip(result, 0, 255)

f1_r = median_filter(f1, kernel_size=3)

f2_r_mean = arithmetic_mean_filter(f2, kernel_size=3)
f2_r_median = median_filter(f2, kernel_size=3)
f2_r_adaptive = adaptive_local_noise_reduction(f2, kernel_size=5, noise_var=200)


f2_r = f2_r_median

# 시각화 비교
plt.figure(figsize=(18, 5))
plt.subplot(141)
plt.imshow(f2, cmap='gray')
plt.title('Original f2')
plt.axis('off')

plt.subplot(142)
plt.imshow(f2_r_mean, cmap='gray')
plt.title('1. Arithmetic Mean')
plt.axis('off')

plt.subplot(143)
plt.imshow(f2_r_median, cmap='gray')
plt.title('2. Median (Selected)')
plt.axis('off')

plt.subplot(144)
plt.imshow(f2_r_adaptive, cmap='gray')
plt.title('3. Adaptive Local')
plt.axis('off')
plt.show()
```

- 복원된 f1, f2를 시각화하기

```python
# f1_r: denoised f1
# f2_r: denoised f2
plt.figure(dpi=128)
plt.subplot(121)
plt.imshow(f1_r,cmap='gray')
plt.axis('off')
plt.subplot(122)
plt.imshow(f2_r,cmap='gray')
plt.axis('off')
plt.show()
```
